```python
#PROYECTO: Consumo de API y almacenamiento en SQLite


#Autor: Daniel Felipe Martínez Silva
#Fecha: 2026

#Descripción:
#Este programa realiza una petición HTTP a una API pública,extrae datos en formato JSON, los almacena en una base de datos SQLite y genera archivos de evidencia en formato CSV y TXT para verificar
#que el proceso se realizó correctamente.

#Tecnologías utilizadas:
#- Requests: para realizar la petición HTTP a la API.
#- SQLite3: para crear y gestionar la base de datos local.
#- Pandas: para manipular los datos y generar archivos CSV.

#API utilizada:
#https://jsonplaceholder.typicode.com/users

#¿que hace?¿para que ?¿como?:
#Este programa realiza una petición HTTP a una API pública,extrae datos en formato JSON, los almacena en una basede datos SQLite y genera archivos de evidencia en CSV y TXT para verificar que el proceso fue exitoso.

import requests # importacion de la libreria Requests para enviar peticion a la API https://jsonplaceholder.typicode.com/users
import sqlite3 # importacion de la interface para conexion y creacion de bases de datos SQLite
import pandas as pd # importacion de Pandas, libreria para el analisis de datos en Python

# Definicion de la URL de la API para envio de peticion y recepcion de datos usando protocolo HTTP
# desde la API https://jsonplaceholder.typicode.com/
API_URL = "https://jsonplaceholder.typicode.com/users"


# Creacion de la funcion extraer_datos validandose codigo 200 (exito en la conexion) o codigo 404 (codigo de error en la conexion)
def extraer_datos(url):
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()
    else:
        raise Exception(f"Error al conectar con la API: {response.status_code}")


# Creacion de la base de datos en SQLite "datos_api.db"
# se realiza la conexion a dicha base de datos
# y se crea el "cursor" para "execute" el CREATE TABLE y el INSERT OR REPLACE INTO usuarios
def guardar_en_db(datos):
    conn = sqlite3.connect('datos_api.db')
    cursor = conn.cursor()

    # Creación de la tabla "usuarios" dentro de la base de datos "datos_api.db"
    # esta contiene los campos id, nombre, usuario y email
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS usuarios (
            id INTEGER PRIMARY KEY,
            nombre TEXT,
            usuario TEXT,
            email TEXT
        )
    ''')

    # Inserción de datos en la tabla "usuarios"(id, nombre, usuario, email)
    for item in datos:
        cursor.execute('''
            INSERT OR REPLACE INTO usuarios (id, nombre, usuario, email)
            VALUES (?, ?, ?, ?)
        ''', (item['id'], item['name'], item['username'], item['email']))

    # Aceptación mediante "Commit" el ingreso de datos en la tabla "usuarios"
    conn.commit()

    # Cerrado de la conexion a la base de datos "datos_api.db"
    conn.close()

# 3. GENERAR EVIDENCIAS (Pandas y Auditoría)
def generar_evidencias(datos_api):
    # --- Archivo Excel/CSV con Pandas ---
    conn = sqlite3.connect('datos_api.db')
    df_db = pd.read_sql_query("SELECT * FROM usuarios", conn)
    df_db.to_csv('muestra_usuarios.csv', index=False)
    conn.close()

    # la variable "registros_api" obtiene la cantidad de registros recibidos desde la
    # API con url "https://jsonplaceholder.typicode.com/users"
    # la variable "registros_db" obtiene la cantidad de registros almacenados en la tabla "usuarios"
    registros_api = len(datos_api)
    registros_db = len(df_db)

    # Creación del archivo "auditoría.txt" con el metodo "open" de Python
    with open('auditoria.txt', 'w', encoding='utf-8') as f:
        f.write("RESUMEN DE AUDITORÍA\n")
        f.write("====================\n")
        f.write(f"Registros extraídos de la API https://jsonplaceholder.typicode.com/: {registros_api}\n")
        f.write(f"Registros guardados en la DB de SQLite3: {registros_db}\n")

        if registros_api == registros_db:
            f.write("\nESTADO: Operacion Exitosa.\n")
        else:
            f.write("\nESTADO: Operacion Fallida.\n")

# EJECUCIÓN DEL PROCESO
try:
    #imprimir el inicio de petición y recepción de datos desde la api
    print("Iniciando proceso de petición y recepción de datos desde la API https://jsonplaceholder.typicode.com/users...")
    datos = extraer_datos(API_URL)

#impimir la base de datos y guardarlos en la tabla
    print("Creando la base de datos datos_api.db y guardando datos en la tabla usuarios...")
    guardar_en_db(datos)
#imprimir archivos de evidencia
    print("Generando archivos de evidencia .CSV y auditoría.txt...")
    generar_evidencias(datos)
#imprimir el proceso ya completado execto si ocurrio un error
    print("¡Proceso completado con éxito!")
except Exception as e:
    print(f"Ocurrió un error: {e}")
    # Si la base de datos ya existe, SQLite no la vuelve a crear,
# simplemente reutiliza el archivo existente.
```

```python

#PROYECTO: Consumo de API y almacenamiento en SQLite

#Autor: Daniel Felipe Martínez Silva
#Fecha: 2026

#¿QUÉ HACE EL PROGRAMA?
#Este programa realiza una petición HTTP a una API pública, extrae datos en
#formato JSON, los almacena en una base de datos SQLite y genera archivos
#de evidencia en CSV y TXT para verificar que el proceso fue exitoso.

#¿PARA QUÉ SIRVE?
#Permite demostrar cómo consumir datos desde una API, almacenarlos en una
#base de datos local y generar archivos de verificación o auditoría.

#¿CÓMO FUNCIONA?
#1. Realiza una petición HTTP a la API.
#2. Recibe los datos en formato JSON.
#3. Guarda los datos en una base de datos SQLite.
#4. Genera un archivo CSV con los datos guardados.
#5. Genera un archivo TXT con un resumen de auditoría del proceso.

#Tecnologías utilizadas:
#- Requests → para consumir la API
#- SQLite3 → para crear y gestionar la base de datos
# Pandas → para manipular datos y generar el archivo CSV

#API utilizada:
https://jsonplaceholder.typicode.com/users

#Importación de librerías necesarias
import requests   # Librería para realizar peticiones HTTP a la API
import sqlite3    # Librería para trabajar con bases de datos SQLite
import pandas as pd   # Librería para análisis y manipulación de datos
# DEFINICIÓN DE LA URL DE LA API

# Esta API pública proporciona información de usuarios
API_URL = "https://jsonplaceholder.typicode.com/users"

# FUNCIÓN: EXTRAER DATOS DE LA API

def extraer_datos(url):
    """
    Realiza una petición HTTP a la API y obtiene los datos.

    Parámetros:
    url (str): Dirección de la API.

    Retorna:
    list: Lista de diccionarios con los datos en formato JSON.

    Si la conexión no es exitosa, se genera una excepción.
    """

    response = requests.get(url)

    # Verifica si la respuesta fue exitosa
    if response.status_code == 200:
        return response.json()
    else:
        raise Exception(f"Error al conectar con la API: {response.status_code}")



# FUNCIÓN: GUARDAR DATOS EN LA BASE DE DATOS

def guardar_en_db(datos):
    """
    Guarda los datos obtenidos desde la API en una base de datos SQLite.

    debe tener:
    datos (list): Lista de diccionarios con información de los usuarios.

    Proceso:
    - Crea la base de datos si no existe.
    - Crea la tabla 'usuarios' si no existe.
    - Inserta los registros obtenidos desde la API.
    """

    # Conexión a la base de datos SQLite
    conn = sqlite3.connect('datos_api.db')
    cursor = conn.cursor()

    # Creación de la tabla usuarios si no existe
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS usuarios (
            id INTEGER PRIMARY KEY,
            nombre TEXT,
            usuario TEXT,
            email TEXT
        )
    ''')

    # Inserción de datos en la tabla
    for item in datos:
        cursor.execute('''
            INSERT OR REPLACE INTO usuarios (id, nombre, usuario, email)
            VALUES (?, ?, ?, ?)
        ''', (item['id'], item['name'], item['username'], item['email']))

    # Guardar cambios en la base de datos
    conn.commit()

    # Cerrar conexión
    conn.close()



# FUNCIÓN: GENERAR ARCHIVOS DE EVIDENCIA
def generar_evidencias(datos_api):
    """
    Genera archivos de evidencia para validar el proceso.

    Archivos generados:
    - muestra_usuarios.csv → contiene los datos guardados en la base de datos
    - auditoria.txt → contiene un resumen del proceso

    También compara:
    - Cantidad de registros obtenidos desde la API
    - Cantidad de registros almacenados en la base de datos
    """

    # Conexión a la base de datos
    conn = sqlite3.connect('datos_api.db')

    # Lectura de los datos desde la tabla usuarios
    df_db = pd.read_sql_query("SELECT * FROM usuarios", conn)

    # Generación del archivo CSV
    df_db.to_csv('muestra_usuarios.csv', index=False)

    # Cerrar conexión
    conn.close()

    # Cantidad de registros obtenidos desde la API
    registros_api = len(datos_api)

    # Cantidad de registros guardados en la base de datos
    registros_db = len(df_db)

    # Creación del archivo de auditoría
    with open('auditoria.txt', 'w', encoding='utf-8') as f:

        f.write("RESUMEN DE AUDITORÍA\n")
        f.write("====================\n")

        f.write(f"Registros extraídos de la API https://jsonplaceholder.typicode.com/: {registros_api}\n")
        f.write(f"Registros guardados en la DB de SQLite3: {registros_db}\n")

        # Verificación de consistencia de datos
        if registros_api == registros_db:
            f.write("\nESTADO: Operación Exitosa.\n")
        else:
            f.write("\nESTADO: Operación Fallida.\n")


# BLOQUE PRINCIPAL DEL PROGRAMA

# Aquí se ejecuta todo el proceso del proyecto o se imprime :
# 1. Se obtienen los datos desde la API
# 2. Se guardan en la base de datos SQLite
# 3. Se generan archivos de evidencia

try:

    # Inicio del proceso
    print("Iniciando proceso de petición y recepción de datos desde la API...")

    # Extracción de datos
    datos = extraer_datos(API_URL)

    # Creación de base de datos y almacenamiento
    print("Creando la base de datos datos_api.db y guardando datos en la tabla usuarios...")
    guardar_en_db(datos)

    # Generación de archivos de evidencia
    print("Generando archivos de evidencia .CSV y auditoria.txt...")
    generar_evidencias(datos)

    # Finalización del proceso
    print("¡Proceso completado con éxito!")

except Exception as e:

    # Manejo de errores
    print(f"Ocurrió un error durante la ejecución: {e}")

# Si la base de datos ya existe, SQLite no la vuelve a crear pues sera inecesario o da error.
# Simplemente buelve a utiliza el archivo existente y continúa
# trabajando con la información almacenada.


# GLOSARIO de terminos:
# API:
# Una API es como un "mesero" en internet. Cuando un programa
# necesita información, le pide al mesero (API) que vaya a
# buscarla y se la traiga. En este proyecto usamos una API
# que nos entrega información de usuarios de ejemplo.

# JSON:
# JSON es una forma de ordenar información para que las
# computadoras puedan entenderla fácilmente. Es como una
# lista organizada con nombres y valores. Por ejemplo:
# nombre = "Juan", edad = 20, correo = "juan@email.com".

# LIBRERÍAS:
# Las librerías son como cajas de herramientas que ya vienen
# con funciones listas para usar. En lugar de crear todo desde
# cero, usamos herramientas que otras personas ya programaron.

# REQUESTS:
# Es una librería que sirve para conectarse a internet desde
# Python y pedir información a páginas web o APIs.

# SQLITE3:
# Es una herramienta que permite crear una base de datos.
# Una base de datos es como un cuaderno donde se guardan
# muchos datos organizados en tablas.

# PANDAS:
# Pandas es una librería que ayuda a organizar los datos
# como si fueran tablas de Excel para poder analizarlos
# o guardarlos en archivos.

# IMPORT:
# La palabra import significa "traer". Se usa para traer
# una librería al programa para poder usar sus herramientas.

# VARIABLE:
# Una variable es como una cajita donde guardamos un dato
# para usarlo después. Por ejemplo, una variable puede
# guardar un número, un texto o una dirección de internet.

# FUNCIÓN:
# Una función es un bloque de código que realiza una tarea
# específica. Es como una pequeña máquina dentro del programa
# que hace algo cuando se le pide.

# PARÁMETRO:
# Un parámetro es la información que se le da a una función
# para que pueda trabajar. Es como darle ingredientes a una
# receta para que se pueda preparar.

# REQUESTS.GET():
# Esta función se usa para enviar una solicitud a una página
# web o API y pedirle información.

# RESPONSE:
# Response significa "respuesta". Es la información que el
# servidor envía después de recibir nuestra solicitud.

# STATUS CODE:
# Es un número que indica si la petición funcionó bien o no.
# Por ejemplo:
# 200 significa que todo salió bien.
# 404 significa que la página no existe.

# IF:
# If significa "si". Se usa para verificar una condición.
# Si algo es verdadero, el programa hace una acción.

# ELSE:
# Else significa "si no". Se usa cuando la condición del
# if no se cumple.

# RETURN:
# Return se usa dentro de una función para devolver un
# resultado al lugar donde se llamó la función.

# EXCEPTION:
# Una excepción es un error que ocurre durante la ejecución
# del programa. Sirve para avisar que algo salió mal.

# BASE DE DATOS:
# Una base de datos es un lugar donde se guarda mucha
# información organizada para poder usarla después.

# SQLITE:
# SQLite es un tipo de base de datos muy simple que guarda
# toda la información en un solo archivo.

# CONEXIÓN A LA BASE DE DATOS:
# Es el proceso de abrir comunicación entre el programa y
# la base de datos para poder guardar o leer información.

# CURSOR:
# El cursor es una herramienta que permite ejecutar
# instrucciones dentro de la base de datos.

# SQL:
# SQL es el lenguaje que se usa para hablar con la base
# de datos y decirle qué hacer.

# CREATE TABLE:
# Es una instrucción que crea una tabla dentro de la
# base de datos donde se guardarán los datos.

# INSERT:
# Es una instrucción que sirve para guardar nuevos datos
# dentro de una tabla de la base de datos.

# FOR:
# El for es un ciclo que repite una acción varias veces.
# Por ejemplo, recorrer todos los usuarios obtenidos
# desde la API.

# COMMIT:
# Commit guarda definitivamente los cambios realizados
# en la base de datos.

# CLOSE:
# Close cierra la conexión con la base de datos cuando
# ya no se necesita usarla.

# DATAFRAME:
# Un DataFrame es como una tabla de Excel dentro de Python,
# con filas y columnas para organizar datos.

# CSV:
# Es un tipo de archivo que guarda datos en forma de tabla
# y que puede abrirse fácilmente en Excel.

# TXT:
# Es un archivo de texto simple que guarda información
# escrita, como un resumen del proceso.

# LEN():
# Len es una función que cuenta cuántos elementos hay
# dentro de una lista o conjunto de datos.

# TRY:
# Try significa "intentar". Se usa para ejecutar un código
# que podría generar un error.

# EXCEPT:
# Except se usa para capturar y manejar el error si algo
# falla dentro del bloque try.

# PRINT:
# Print muestra mensajes en la consola para que el usuario
# pueda ver lo que está haciendo el programa.
```
